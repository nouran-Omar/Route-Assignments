## 1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.

**Input:** `list=["apple", "pie", "banana"], K=3`

**Output:** `['apple', 'banana']`

```python
def filter_long_strings(strings, k):
    """
    Filter and return strings whose length is strictly greater than k.

    Args:
        strings (list[str]): List of string elements to filter.
        k (int): Minimum length threshold (exclusive).

    Returns:
        list[str]: New list containing only strings with len(s) > k.
    """
    # Use a list comprehension to keep only elements longer than the threshold k
    return [s for s in strings if len(s) > k]


# Display the function's docstring and a visual separator
print(filter_long_strings.__doc__)
print("-" * 60)

# Collect raw comma-separated words and the length threshold K outside the function
user_input = input("Enter strings separated by commas: ")
k_input = int(input("Enter number K: "))

# Clean leading/trailing whitespace around each item before filtering
strings_list = [s.strip() for s in user_input.split(",")]

# Call the function with explicit arguments and print the resulting list
print("Output:", filter_long_strings(strings_list, k_input))
```

### 2. Write a function named area() that calculates the area of either a "circle" or a "rectangle". The function should take the shape name as the first argument, followed by the necessary dimensions (radius for circle; length and width for rectangle). 
**Example Input:**`area("circle", 5) OR area("rectangle", 4, 6) `

**Example Output:**` Circle Area: 78.5, Rectangle Area: 24`

```python
def area(shape, *dims):
    """
    Calculate the area of a circle or rectangle from a shape name and dimensions.

    Args:
        shape (str): Target shape name ("circle" or "rectangle").
        *dims: Variable-length numeric dimensions (radius for circle;
               length and width for rectangle).

    Returns:
        str: Formatted area string, or a message if the shape is unsupported.
    """
    # Normalize shape name to lowercase for case-insensitive matching
    shape_normalized = shape.strip().lower()

    if shape_normalized == "circle":
        # Extract the radius from the first positional argument in *dims
        radius = dims[0]
        # Use 3.14 for pi as expected to match 78.5 for radius 5
        result = 3.14 * (radius ** 2)
        return f"Circle Area: {result}"

    elif shape_normalized == "rectangle":
        # Extract length and width from *dims and multiply them
        length = dims[0]
        width = dims[1]
        result = length * width
        return f"Rectangle Area: {result}"

    return "Unsupported shape. Please choose 'circle' or 'rectangle'."


# Display the function's docstring and a visual separator
print(area.__doc__)
print("-" * 60)

# Prompt the user for the shape type and the required dimensions outside the function
user_shape = input("Enter shape name (circle or rectangle): ").strip()

if user_shape.lower() == "circle":
    # Read radius as a float to support decimal or integer radii
    r = float(input("Enter radius: "))
    print(area(user_shape, r))

elif user_shape.lower() == "rectangle":
    # Read length and width as integers so whole-number inputs produce integer areas (e.g., 24)
    l = int(input("Enter length: "))
    w = int(input("Enter width: "))
    print(area(user_shape, l, w))

else:
    print("Invalid shape name entered.")
```

## Write a program that takes an email address as input and splits it into two parts: the "username" and the "domain". 
**Example Input:** `support@google.com  `

**Example Output:** `Username: support, Domain: google.com`

```python
def separate_email(email: str):
    """
Takes an email address as a string, splits it at the '@' symbol,
and returns the username and domain parts separately.
    """
    email_parts = email.split("@")
    username = email_parts[0]
    domain = email_parts[1]
    return username, domain

print(separate_email.__doc__)
print("-" *60)

user_email = input("Enter the email: ")
user_name, user_domain = separate_email(user_email)

print(f"Output: Username: {user_name}, Domain: {user_domain}")
```

### 4. Write a program to find the sum of all numbers in a list. ensure the program ignores any elements in the list that are not numbers (like strings or booleans). 

**Example Input:**

`[10, "apple", 20, True, 5] `

**Example Output:** 
`Total Sum: 35`

```python
def find_sum_of_numbers(numbers_list):
    """
Takes a list containing mixed data types, iterates through it,
and calculates the total sum of numeric elements only 
while ignoring strings and boolean values.
    """
    total_sum = 0
    for item in numbers_list:
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            total_sum += item
    return total_sum
print(find_sum_of_numbers.__doc__)
print("-" * 60)
input_numbers = [10, "apple", 20, True, 5]
result = find_sum_of_numbers(input_numbers)
print(f"Output: Total Sum: {result}")
```

### 5. Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.) Perform the following tasks now: 
<input type="checkbox" checked>  *Now add items to the menu.* 

<input type="checkbox" checked>  *Make table reservations.* 

<input type="checkbox" checked>  *Take customer orders.*

<input type="checkbox" checked>  *Print the menu.* 

<input type="checkbox" checked>  *Print table reservations.* 

<input type="checkbox" checked>  *Print customer orders.*

```python
class Restaurant:
    def __init__(self):
        self.menu_items = []
        self.book_table = {}
        self.customer_orders = {}

    def add_item_to_menu(self, item):
        self.menu_items.append(item)

    def book_tables(self, table_number, customer_name):
        self.book_table[table_number] = customer_name

    def customer_order(self, customer_name, item):
        if customer_name not in self.customer_orders:
            self.customer_orders[customer_name] = []
        self.customer_orders[customer_name].append(item)

    def get_menu(self):
        return self.menu_items

    def get_table(self):
        return self.book_table

    def get_customer_order(self):
        return self.customer_orders

my_restaurant = Restaurant()

my_restaurant.add_item_to_menu("Pizza")
my_restaurant.add_item_to_menu("Burger")
my_restaurant.add_item_to_menu("Pasta")

my_restaurant.book_tables(1, "Nouran")
my_restaurant.book_tables(2, "Ahmed")

my_restaurant.customer_order("Nouran", "Pizza")
my_restaurant.customer_order("Ahmed", "Burger")
my_restaurant.customer_order("Nouran", "Burger")
my_restaurant.customer_order("Nouran", "Pasta")

print(f"Menu Items: {my_restaurant.get_menu()}")
print(f"Table Reservations: {my_restaurant.get_table()}")
print(f"Customer Orders: {my_restaurant.get_customer_order()}")
```

### class Bike that is composed of the following attributes:
* <input type="checkbox" checked>  name
* <input type="checkbox" checked>   price

Here are some of the methods that you need to add to your class Bike;
* <input type="checkbox" checked>  def __init__(self)
* <input type="checkbox" checked>  Getter / setter for all attributes.

```python
class Customer:
    """Represents a customer renting bikes from the shop."""

    def __init__(self, name="", age=0):
        # Encapsulate customer attributes with private naming convention
        self.__name = name
        self.__age = age

    def get_name(self):
        """Return the customer's name."""
        return self.__name

    def set_name(self, name):
        """Update the customer's name."""
        self.__name = name

    def get_age(self):
        """Return the customer's age."""
        return self.__age

    def set_age(self, age):
        """Update the customer's age."""
        self.__age = age


class Bike:
    """Represents an individual bike available in the rental store."""

    def __init__(self, name="", price=0.0):
        # Encapsulate bike attributes with private naming convention
        self.__name = name
        self.__price = price

    def get_name(self):
        """Return the bike's name."""
        return self.__name

    def set_name(self, name):
        """Update the bike's name."""
        self.__name = name

    def get_price(self):
        """Return the bike's price/value."""
        return self.__price

    def set_price(self, price):
        """Update the bike's price/value."""
        self.__price = price


class MainBikeRental:
    """Manages bike inventory, active rentals, billing, and returning bikes."""

    def __init__(self, bikes=None, rentals=None):
        # Initialize separate lists per instance to avoid mutable default argument bugs
        self.__bikes = list(bikes) if bikes is not None else []
        self.__rentals = list(rentals) if rentals is not None else []

    def get_bikes(self):
        """Return the list of bikes currently available in stock."""
        return self.__bikes

    def set_bikes(self, bikes):
        """Replace the list of bikes currently in stock."""
        self.__bikes = bikes

    def get_rentals(self):
        """Return the list of active rental records."""
        return self.__rentals

    def set_rentals(self, rentals):
        """Replace the list of active rental records."""
        self.__rentals = rentals

    def addBike(self, bike):
        """Add a Bike instance to the store's available inventory."""
        self.__bikes.append(bike)

    def totalCost(self):
        """Calculate and return the total value/cost of bikes currently in the store."""
        total = 0
        for bike in self.__bikes:
            total += bike.get_price()
        return total

    def requestBike(self, customer, num_of_bikes):
        """
        Process a bike rental request if the customer is older than 6
        and enough bikes are available in stock.
        """
        # Enforce minimum age requirement (must be strictly greater than 6)
        if customer.get_age() <= 6:
            print("Sorry, the customer must be older than 6 years to rent a bike.")
            return False

        # Check stock availability before deducting bikes
        if num_of_bikes <= len(self.__bikes):
            rented_bikes = []
            for _ in range(num_of_bikes):
                # Pop from the front of the stock list to preserve order
                rented_bikes.append(self.__bikes.pop(0))

            # Record the rental along with the actual Bike objects rented
            rental_info = {"customer": customer, "bikes": rented_bikes}
            self.__rentals.append(rental_info)

            print(f"Success! {num_of_bikes} bike(s) rented to {customer.get_name()}.")
            return True
        else:
            # Inform the customer to wait when requested quantity exceeds stock
            print("Wait 10 mins.")
            return False

    def returnBike(self, rentalTime, num_of_bikes):
        """
        Calculate the rental bill (rentalTime * 40) and restore the actual
        returned Bike objects from the latest rental record back into stock.
        """
        # Compute total bill at 40 per unit of rental time
        total_bill = rentalTime * 40
        print(f"Total bill cost: {total_bill} EGP.")

        # Restore original Bike objects from the most recent rental entry
        if self.__rentals:
            last_rental = self.__rentals[-1]
            rented_bikes = last_rental["bikes"]

            # Take up to num_of_bikes from the rental record and return them to stock
            bikes_to_return = rented_bikes[:num_of_bikes]
            self.__bikes.extend(bikes_to_return)
            last_rental["bikes"] = rented_bikes[num_of_bikes:]

            # Remove the rental record once all of its bikes have been returned
            if not last_rental["bikes"]:
                self.__rentals.pop()

            print(f"Successfully returned {len(bikes_to_return)} bike(s) back to the store stock.")
        else:
            print("No active rental records found.")

        return total_bill


# --- Example Usage & Verification ---
shop = MainBikeRental()

# Add initial bikes to store stock (total value = 150 + 200 + 100 = 450)
shop.addBike(Bike("Mountain Bike", 150))
shop.addBike(Bike("Road Bike", 200))
shop.addBike(Bike("City Bike", 100))

print("Total cost of bikes in store:", shop.totalCost())

# Create a customer and rent 2 bikes
customer1 = Customer("Nouran", 20)
shop.requestBike(customer1, 2)

# Return 2 bikes after 3 rental time units (bill = 3 * 40 = 120)
shop.returnBike(rentalTime=3, num_of_bikes=2)

# Verify that the original bikes are restored and total store cost is back to 450
print("Total cost of bikes in store after return:", shop.totalCost())
```
