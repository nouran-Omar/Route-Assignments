## 1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.

**Input:** `list=["apple", "pie", "banana"], K=3`

**Output:** `['apple', 'banana']`

```python
def filter_long_strings():
    user_input = input("Enter strings separated by commas: ")
    k_input = input("Enter number K: ")
    
    strings = []
    for s in user_input.split(","):
        strings.append(s.strip())
        
    k = int(k_input)
    
    result = []
    for word in strings:
        if len(word) > k:
            result.append(word)
            
    print(f"Input: list = {strings}, K = {k}")
    return result
print(f"Output: {filter_long_strings()}")
```

### 2. Write a function named area() that calculates the area of either a "circle" or a "rectangle". The function should take the shape name as the first argument, followed by the necessary dimensions (radius for circle; length and width for rectangle). 
**Example Input:**`area("circle", 5) OR area("rectangle", 4, 6) `

**Example Output:**` Circle Area: 78.5, Rectangle Area: 24`

```python
def area(shape, *dims):
    if shape.lower() == "circle":
        radius = dims[0]
        result = 3.14 * (radius ** 2)
        return f"Circle Area: {result}"
    elif shape.lower() == "rectangle":
        length = dims[0]
        width = dims[1]
        result = length * width
        return f"Rectangle Area: {result}"

user_shape = input("Enter shape name (circle or rectangle): ")

if user_shape.lower() == "circle":
    r = float(input("Enter radius: "))
    print(area(user_shape, r))
elif user_shape.lower() == "rectangle":
    l = float(input("Enter length: "))
    w = float(input("Enter width: "))
    print(area(user_shape, l, w))
```

```python
import math

def area(shape_name):
    calculated_area = 0
    if shape_name.lower() == 'circle':
        r = float(input("Please enter the radius: "))
        calculated_area = r * r * math.pi
    else:
        w = float(input("Please enter the width: "))
        l = float(input("Please enter the length: "))
        calculated_area = l * w
    return calculated_area

shape_name = input("""
Calculates the area of either a "circle" or a 
"rectangle": 
""")

print(f"Output: {shape_name} Area: {area(shape_name):.2f}")
```

## Write a program that takes an email address as input and splits it into two parts: the "username" and the "domain". 
**Example Input:** `support@google.com  `

**Example Output:** `Username: support, Domain: google.com`

```python
def separate_email(email: str):
    email_parts = email.split("@")
    username = email_parts[0]
    domain = email_parts[1]
    return username, domain

user_email = input("Enter the email: ")
user_name, user_domain = separate_email(user_email)

print(f"Output: Username: {user_name}, Domain: {user_domain}")
```

### 4. Write a program to find the sum of all numbers in a list. ensure the program ignores any elements in the list that are not numbers (like strings or booleans). 

**Example Input:**

`[10, "apple", 20, True, 5] `

**Example Output:** 
`Total Sum: 35`

```python
def find_sum_of_numbers(numbers_list):
    total_sum = 0
    for item in numbers_list:
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            total_sum += item
    return total_sum

input_numbers = [10, "apple", 20, True, 5] 
result = find_sum_of_numbers(input_numbers)

print(f"Input: {input_numbers}")
print(f"Output: Total Sum: {result}")
```

### 5. Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.) Perform the following tasks now: 
<input type="checkbox" checked>  *Now add items to the menu.* 

<input type="checkbox" checked>  *Make table reservations.* 

<input type="checkbox" checked>  *Take customer orders.*

<input type="checkbox" checked>  *Print the menu.* 

<input type="checkbox" checked>  *Print table reservations.* 

<input type="checkbox" checked>  *Print customer orders.*

```python
class Restaurant:
    def __init__(self):
        self.menu_items = []
        self.book_table = {}
        self.customer_orders = {}

    def add_item_to_menu(self, item):
        self.menu_items.append(item)

    def book_tables(self, table_number, customer_name):
        self.book_table[table_number] = customer_name

    def customer_order(self, customer_name, item):
        if customer_name not in self.customer_orders:
            self.customer_orders[customer_name] = []
        self.customer_orders[customer_name].append(item)

    def get_menu(self):
        return self.menu_items

    def get_table(self):
        return self.book_table

    def get_customer_order(self):
        return self.customer_orders

my_restaurant = Restaurant()

my_restaurant.add_item_to_menu("Pizza")
my_restaurant.add_item_to_menu("Burger")
my_restaurant.add_item_to_menu("Pasta")

my_restaurant.book_tables(1, "Nouran")
my_restaurant.book_tables(2, "Ahmed")

my_restaurant.customer_order("Nouran", "Pizza")
my_restaurant.customer_order("Ahmed", "Burger")
my_restaurant.customer_order("Nouran", "Burger")
my_restaurant.customer_order("Nouran", "Pasta")

print(f"Menu Items: {my_restaurant.get_menu()}")
print(f"Table Reservations: {my_restaurant.get_table()}")
print(f"Customer Orders: {my_restaurant.get_customer_order()}")
```

### class Bike that is composed of the following attributes:
* <input type="checkbox" checked>  name
* <input type="checkbox" checked>   price

Here are some of the methods that you need to add to your class Bike;
* <input type="checkbox" checked>  def __init__(self)
* <input type="checkbox" checked>  Getter / setter for all attributes.

```python
class Bike:
    def __init__(self, name="", price=0.0):
        self.name = name
        self.price = price

    def getName(self):
        return self.name

    def getPrice(self):
        return self.price

    def setName(self, name):
        self.name = name

    def setPrice(self, price):
        self.price = price


class Customer:
    def __init__(self, name="", age=0):
        self.name = name
        self.age = age

    def getName(self):
        return self.name

    def getAge(self):
        return self.age

    def setName(self, name):
        self.name = name

    def setAge(self, age):
        self.age = age


class MainBikeRental:
    def __init__(self, bikes=None, rentals=None):
        self.bikes = bikes if bikes is not None else []
        self.rentals = rentals if rentals is not None else []

    def getBikes(self):
        return self.bikes

    def setBikes(self, bikes):
        self.bikes = bikes

    def getRentals(self):
        return self.rentals

    def setRentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, num_of_bikes):
        if customer.getAge() <= 6:
            print("Wait 10 mins (Age must be over 6 years)")
            return False
        
        if len(self.bikes) >= num_of_bikes:
            rented_batch = []
            for _ in range(num_of_bikes):
                rented_batch.append(self.bikes.pop(0))
            self.rentals.append({"customer": customer, "bikes": rented_batch})
            print(f"Success! {num_of_bikes} bike(s) rented.")
            return True
        else:
            print("Wait 10 mins")
            return False

    def returnBike(self, rentalTime, num_of_bikes, returned_bikes):
        total_bill = rentalTime * 40
        for bike in returned_bikes:
            self.bikes.append(bike)
        print(f"Total Bill Cost: {total_bill}")
        return total_bill

    def totalCost(self):
        return sum(bike.getPrice() for bike in self.bikes)



rental_shop = MainBikeRental()

bike1 = Bike("Giant", 1500.0)
bike2 = Bike("Trek", 2000.0)
bike3 = Bike("Merida", 1800.0)

rental_shop.addBike(bike1)
rental_shop.addBike(bike2)
rental_shop.addBike(bike3)


customer1 = Customer("Nouran", 21)

print("--- Requesting Bikes ---")
rental_shop.requestBike(customer1, 2)

print("\n--- Returning Bikes ---")
returned_batch = [bike1, bike2]
rental_shop.returnBike(2, 2, returned_batch)

print(f"\nTotal price of remaining bikes in store: {rental_shop.totalCost()}")
```
