## 1. Ask the user for a string that contains both letters and numbers (e.g., `"Py2th0on3"`). 

Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word.

* **Example Input:** `"Py2th0on3"`
* **Example Output:** `Word: "Python", Sum: 5`

````python
def parse_and_sum_mixed_string(text: str):
    """
    Helper function to separate letters and numbers from a string,
    join letters into a word, and return the sum of numbers.
    """
    str_list = []
    num_list = []

    for char in text:
        if char.isdigit():
            num_list.append(int(char))
        else:
            str_list.append(char)

    word = "".join(str_list)
    total_sum = sum(num_list)

    return word, total_sum


user_input = input("Enter string that contains both letters and numbers ? ")

# 2. Call the  function
word, total_sum = parse_and_sum_mixed_string(user_input)


print('sum =', total_sum)
print('Word :', word)
````

## 2. Ask a user to enter a password. Check if it is valid based on these rules: 
* Length must be between 8 and 16 characters.
* It must contain at least one uppercase letter and one special character (e.g., `@`, `#`, `$`). 

* **Example Input:** `"Guest@123"`
* **Example Output:** `"Password is Secure."`


````python
def check_password_security(password: str):
    """
    Helper function to validate password security based on length, 
    uppercase letters, and special characters.
    """
    symbols = "~!@#$%^&*?_"
    score = 0

    if 8 <= len(password) <= 16:
        has_upper = False
        has_symbol = False
        
        for i in password:
            if i.isupper():
                has_upper = True
            elif i in symbols:
                has_symbol = True
                
        if has_upper:
            score += 1
        if has_symbol:
            score += 1
            
        if score == 2:
            return "Password is Secure."
        else:
            return "Not secure."
    else:
        return "Not secure: Length must be between 8 and 16 characters."

user_password = input("Enter your password: ")
result = check_password_security(user_password)
print(result)
````

## 3. A store gives discounts based on the total price: 
* **Under $50:** No discount
* **$50 to $100:** 10% discount
* **Over $100:** 20% discount

If the user is a **"Member"**, they get an additional **5% off** the final calculated price.

* **Example Input:** `Price: 100`, `Member: Yes`
* **Example Output:** `Final Total: $85.5` *(Note: 100 - 10% = 90; then 90 - 5% = 85.5)*

````python
def calculate_final_price(price: float, has_member: bool) -> float:
    """
    Helper function to calculate the final price after applying 
    store discounts based on the total price and membership status.
    """
    total = price
    
    if 50 <= price <= 100:
        total = price - (price * 0.10)  # 10% discount
    elif price > 100:
        total = price - (price * 0.20)  # 20% discount
    else:
        total = price                   # No discount (< $50)

    if has_member:
        total = total - (total * 0.05)  # Additional 5% off

    return total

price = float(input("Enter the total price: "))
has_member_input = input("Do You Member (Yes OR No)? ").strip().capitalize() == "Yes"

final_total = calculate_final_price(price, has_member_input)

print(f"Final Total: ${final_total}")
````

5. ### Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the average for each student. If the average is > 90, print the student's name followed by "Honor Roll."

* **Example Input:** `{'Sam': [95, 90, 92]}`
* **Example Output:** `Sam: Average 92.3 - Honor Roll`


````python
def evaluate_student_scores (name, scores_input):
    """
    Helper function to calculate student average score and determine 
    if they qualify for the Honor Roll (average > 90).
    """
    students_scores = {}
    
    scores_list_raw = scores_input.split(',')
    scores = []

    for item in scores_list_raw:
        clean_item = item.strip()
        if clean_item:
            float_item = float(clean_item)
            scores.append(float_item)

    students_scores[name] = scores

    for student_name, student_scores in students_scores.items():
        total_score = 0.0
        for s in student_scores:
            total_score += s
            
        count_scores = len(student_scores)
        average = total_score / count_scores
        
        if average > 90:
            return f"{student_name}: Average {average:.1f} - Honor Roll"
        else:
            return f"{student_name}: Average {average:.1f}"


name_input = input("Enter the student name: ")
scores_input = input("Enter 3 test scores separated by comma (e.g. 95,90,92): ")

score = evaluate_student_scores(name_input, scores_input)
print(score)
````

### 6. Ask the user for two words of the same length. Create a new string by interleaving their characters (one from the first word, one from the second).

* **Example Input:** `word1 = "abc"`, `word2 = "xyz"`
* **Example Output:** `"axbycz"`


````python
def interleave_strings():
    """
    Helper function to ask the user for two words of the same length
    and interleave their characters together.
    """
    while True:
        string_one = input("Enter string one: ").strip()
        string_two = input("Enter string two (must be same length as string one): ").strip()
        
        if len(string_one) == len(string_two):
            string_third = ""
            for i in range(len(string_one)):
                string_third += (string_one[i] + string_two[i])
            return string_third
        else:
            print("Lengths are not equal! Please try again.\n")



result = interleave_strings()
print(result)
````

## 7. Ask for a multi-digit number. Sum its digits. If the result is more than one digit, keep summing the digits until you get a single-digit result.

* **Example Input:** `98`
* **Example Output:** `Digital Root: 8 (9+8=17, 1+7=8)`
```[cite: 1]


````python
def calculate_digital_root():
    num_one = input("Enter a multi-digit number: ").strip()
    while num_one.isdigit() and int(num_one) >= 10:
        current_sum = 0
        for char in num_one:
            current_sum += int(char)
        num_one = str(current_sum)
    return f"Digital Root: {num_one}"

result = calculate_digital_root()
print(result)
````

## 8. Given a list of numbers with many duplicates, return a list of tuples showing each unique element and its frequency (how many times it appeared).

* **Example Input:** `[1, 1, 2, 3, 3, 3, 4]`
* **Example Output:** `[(1, 2), (2, 1), (3, 3), (4, 1)]`


````python
def count_element_frequency():
    numbers = input("Enter numbers separated by comma: ").split(",")
    numbers_list = [int(item.strip()) for item in numbers if item.strip()]
    
    counts = {}
    for num in numbers_list:
        counts[num] = counts.get(num, 0) + 1
        
    result_list = list(counts.items())
    return result_list

result = count_element_frequency()
print(result)
````

## 9. Ask the user for a number $n$. Calculate the factorial ($n!$) using a loop, and then check if the resulting factorial is divisible by the sum of numbers from 1 to $n$.

* **Example Input:** `4`
* **Example Output:** `Factorial is 24, Sum is 10. 24 is not divisible by 10.`

````python
def check_factorial_divisibility():
    n = int(input("Enter a number n: ").strip())
    

    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
        
    total_sum = 0
    for i in range(1, n + 1):
        total_sum += i
        

    if total_sum != 0 and factorial % total_sum == 0:
        return f"Factorial is {factorial}, Sum is {total_sum}. {factorial} is divisible by {total_sum}."
    else:
        return f"Factorial is {factorial}, Sum is {total_sum}. {factorial} is not divisible by {total_sum}."

result = check_factorial_divisibility()
print(result)
````

## 10. Start with a dictionary of items and prices: {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}. Allow the user to "Add" or "Remove" items from a cart. If they add an item not in the store, print an error. Finally, print the total bill.

* **Example Input:** `Add 'apple', Add 'milk'`
* **Example Output:** `Total Bill: $2.0`

````python
def manage_store_cart():
    dictionary_of_items = {
        'apple': 0.5,
        'bread': 2.0,
        'milk': 1.5
    }
    
    print("""Store Items and Prices:
apple: 0.5
bread: 2.0
milk: 1.5
""")
    
    total_bill = 0.0
    while True:
        action = input("Do you want to 'Add', 'Remove', or type 'done' to finish? ").strip().lower()
        
        if action == 'done':
            break
            
        elif action == 'add':
            item_name = input("Enter item name to add: ").strip().lower()
            if item_name in dictionary_of_items:
                total_bill += dictionary_of_items[item_name]
                print(f"Added {item_name}. Current Total: ${total_bill}")
            else:
                print("Error: Item not found in the store.")
                
        elif action == 'remove':
            item_name = input("Enter item name to remove: ").strip().lower()
            if item_name in dictionary_of_items:
                total_bill -= dictionary_of_items[item_name]
                print(f"Removed {item_name}. Current Total: ${total_bill}")
            else:
                print("Error: Item not found in the store.")
                
        else:
            print("Invalid choice, please type 'Add', 'Remove', or 'done'.")
            
    return f"Total Bill: ${total_bill}"

result = manage_store_cart()
print(result)
````

## 11. Check if a full sentence is a palindrome, ignoring spaces and capitalization.

* **Example Input:** `"Race car"`
* **Example Output:** `True`



````python
def check_sentence_palindrome():
    sentence = input("Enter a sentence to check if it's a palindrome: ")
    cleaned = "".join(sentence.split()).lower()
    return cleaned == cleaned[::-1]

result = check_sentence_palindrome()
print(result)
````

### 13. In a list of numbers, find all "peaks." A peak is a number that is greater than the number to its left and the number to its right. 
**Example Input:**`[1, 5, 2, 6, 3]` 
**Example Output:** `Peaks: [5, 6] `


````python
def find_peaks():
    user_input = input("Enter numbers separated by comma (e.g. 1,5,2,6,3): ")
    inp = [int(item.strip()) for item in user_input.split(",") if item.strip()]
    
    peaks = []
    for i in range(1, len(inp) - 1):
        if inp[i] > inp[i - 1] and inp[i] > inp[i + 1]:
            peaks.append(inp[i])
            
    return f"Peaks: {peaks}"

result = find_peaks()
print(result)
````

### 14. Take a dictionary like {'A': 1, 'B': 2, 'C': 1} and flip it. If two keys have the same value, the new value 
## should be a list of those keys. 
 *Example Input:* `{'A': 1, 'B': 2, 'C': 1} `
 *Example Output:* `{1: ['A', 'C'], 2: ['B']} `


````python
def flip_dictionary():
    original_dict = {'A': 1, 'B': 2, 'C': 1}
    flipped_dict = {}
    
    for key, value in original_dict.items():
        if value not in flipped_dict:
            flipped_dict[value] = []
        flipped_dict[value].append(key)
        
    return flipped_dict

result = flip_dictionary()
print(result)
````

### 15. Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear. 
**Example Input:**` "banana" `

**Example Output:** `{'a': [1, 3, 5]}`

````python
def find_vowel_indices():
    text = input("Enter a string: ")
    vowels = "aeiou"
    vowels_dic = {}
    
    for i in range(len(text)):
        char = text[i].lower()
        
        if char in vowels:
            if char not in vowels_dic:
                vowels_dic[char] = []
                
            vowels_dic[char].append(i)
            
    return vowels_dic

result = find_vowel_indices()
print(result)
````
