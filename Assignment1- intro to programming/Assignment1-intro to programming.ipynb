## 1. Ask the user for a string that contains both letters and numbers (e.g., `"Py2th0on3"`). 

Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word.

* **Example Input:** `"Py2th0on3"`
* **Example Output:** `Word: "Python", Sum: 5`

```python
def parse_and_sum_mixed_string(text: str):
    """
    Helper function to separate letters and numbers from a string,
    join letters into a word, and return the sum of numbers.
    """
    str_list = []
    num_list = []

    for char in text:
        if char.isdigit():
            num_list.append(int(char))
        else:
            str_list.append(char)

    word = "".join(str_list)
    total_sum = sum(num_list)

    return word, total_sum


user_input = input("Enter string that contains both letters and numbers ? ")

# 2. Call the  function
word, total_sum = parse_and_sum_mixed_string(user_input)


print('sum =', total_sum)
print('Word :', word)
```

## 2. Ask a user to enter a password. Check if it is valid based on these rules: 
* Length must be between 8 and 16 characters.
* It must contain at least one uppercase letter and one special character (e.g., `@`, `#`, `$`). 

* **Example Input:** `"Guest@123"`
* **Example Output:** `"Password is Secure."`


```python
def check_password_security(password: str):
    """
    Helper function to validate password security based on length, 
    uppercase letters, and special characters.
    """
    symbols = "~!@#$%^&*?_"
    score = 0

    if 8 <= len(password) <= 16:
        has_upper = False
        has_symbol = False
        
        for i in password:
            if i.isupper():
                has_upper = True
            elif i in symbols:
                has_symbol = True
                
        if has_upper:
            score += 1
        if has_symbol:
            score += 1
            
        if score == 2:
            return "Password is Secure."
        else:
            return "Not secure."
    else:
        return "Not secure: Length must be between 8 and 16 characters."

user_password = input("Enter your password: ")
result = check_password_security(user_password)
print(result)
```

## 3. A store gives discounts based on the total price: 
* **Under $50:** No discount
* **$50 to $100:** 10% discount
* **Over $100:** 20% discount

If the user is a **"Member"**, they get an additional **5% off** the final calculated price.

* **Example Input:** `Price: 100`, `Member: Yes`
* **Example Output:** `Final Total: $85.5` *(Note: 100 - 10% = 90; then 90 - 5% = 85.5)*

```python
def calculate_final_price(price: float, has_member: bool) -> float:
    """
    Helper function to calculate the final price after applying 
    store discounts based on the total price and membership status.
    """
    total = price
    
    if 50 <= price <= 100:
        total = price - (price * 0.10)  # 10% discount
    elif price > 100:
        total = price - (price * 0.20)  # 20% discount
    else:
        total = price                   # No discount (< $50)

    if has_member:
        total = total - (total * 0.05)  # Additional 5% off

    return total

price = float(input("Enter the total price: "))
has_member_input = input("Do You Member (Yes OR No)? ").strip().capitalize() == "Yes"

final_total = calculate_final_price(price, has_member_input)

print(f"Final Total: ${final_total}")
```
