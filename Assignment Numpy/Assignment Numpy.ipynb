1. Create a 3x3 array with random integers between 1 and 100. Write a program to "normalize" the array so that the values are scaled between 0 and 1.

- Formula: (x - min) / (max - min)

- Example Output: An array where the smallest value becomes 0.0 and the largest becomes 1.0. 

```python
import numpy as np 
```

```python
# Create 3x3 matrix & display
arr = np.random.randint(1, 101, size=(3, 3))
print("Original:\n", arr)

min_val = arr.min()
max_val = arr.max()
rng = max_val - min_val

# Compute normalized values between 0 and 1
normalized = (arr - min_val) / rng if rng != 0 else np.zeros_like(arr, dtype=float)
print("Normalized:\n", normalized)
print("--"*50)
print("--"*50)
```

2. Create a 1D array of 10 random numbers. Ask the user for a value X Find the value 
in the array that is closest to X 

- Example Input: Array: [1, 5, 8, 12], X = 10 

- Example Output: 8 or 12 (whichever is mathematically closer).

```python
# Create a 1D array of 10 random int between 1 and 50
arr = np.random.randint(1, 51, size=10)
print("1D Random Array:", arr)

# Ask user for value X
x = int(input("Enter value X: "))

# Calculate absolute differences 
diffs = np.abs(arr - x)
min_diff = diffs.min()
closest_values = arr[diffs == min_diff]
print("Closest value(s) to", x, "is:", closest_values)

print("--"*50)
print("--"*50)
```

3. Write a NumPy program to create a 5x5 array with random values. For each column 
in the array, find the second-largest value. 

- Example Input:  [[1, 2, 3], [4, 5, 6], [7, 8, 9]] 

- Example Output: [4, 5, 6] (These are the second-highest values for each 
respective column).


```python
import numpy as np

# Create a 5x5 array random 
arr = np.random.randint(1, 101, size=(5, 5))
print("5x5 Random Array:\n", arr)

# Sort each column & select the second-largest row
second_largest_per_col = np.sort(arr, axis=0)[-2]
print("\nSecond-largest value in each column:", second_largest_per_col)
print("--"*50)
print("--"*50)
```

4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an 
even number. Print the indices of these rows.

```python
matrix = np.random.randint(1, 101, size=(6, 4))
print("6x4 Random Matrix:\n", matrix)

# Calculate the sum of elements along each row (axis=1 for rows, axis=0 for columns)
row_sums = np.sum(matrix, axis=1)
print("\nRow sums:", row_sums)

# Identify and print the indices of rows whose sum is even
even_sum_row_indices = np.where(row_sums % 2 == 0)[0]
print("\nIndices of rows with an even sum:", even_sum_row_indices)
print("--"*50)
print("--"*50)
```

5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers 
that are multiples of 3 or 5 with the value -1. 

- Example Input: [3, 7, 10, 15] 

- Example Output: [-1, 7, -1, -1]

```python
# Create a 1D array of 30 random integers
arr = np.random.randint(1, 51, size=30)
print("Original 1D Array:\n", arr)

# Replace elements that are multiples of 3 or 5 with -1
arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1
print("\nModified Array:\n", arr)
print("--"*50)
print("--"*50)
```

6. Write a NumPy program to create a 3x3 array with random values. For each row, 
calculate its mean (average) and subtract that mean from every element in that 
specific row. 

- Example Input: [[10, 20, 30], [5, 5, 5], [0, 10, 20]] 

- Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]

```python
# Create a 3x3 array with random integer
arr = np.random.randint(1, 51, size=(3, 3))
print("3x3 Random Array:\n", arr)

# Calculate the mean for each row while preserving dimensions for broadcasting
row_means = np.mean(arr, axis=1, keepdims=True)
print("\nRow means:\n", row_means)

# Subtract each row's mean from every element in that specific row
centered_arr = arr - row_means
print("\nResult after subtracting row mean:\n", centered_arr)

print("--"*50)
print("--"*50)
```
